```python
query="""

SELECT users.name,COUNT(*) AS total
FROM complaints
LEFT JOIN users ON complaints.user_email=users.email
WHERE complaints.status="pending"
GROUP BY users.name

"""

for row in cur.execute(query):
  print(row)
```
